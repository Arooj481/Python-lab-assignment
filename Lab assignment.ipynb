```python


def calculate_result(python, ai, maths):
    total = python + ai + maths
    per = total / 300 * 100

    if per >= 80:
        grade = "A"
    elif per >= 70:
        grade = "B"
    elif per >= 60:
        grade = "C"
    elif per >= 50:
        grade = "D"
    else:
        grade = "F"

    return total, per, grade

students_list = []

for i in range(5):
    print("\nStudent", i+1, "details:")
    name = input("Enter name: ")
    roll = input("Enter roll no: ")
    py = int(input("Enter Python marks: "))
    ai = int(input("Enter AI marks: "))
    math = int(input("Enter Maths marks: "))

    total, per, grade = calculate_result(py, ai, math)

    student = {
        "name": name,
        "roll": roll,
        "python": py,
        "ai": ai,
        "maths": math,
        "total": total,
        "percentage": per,
        "grade": grade
    }

    students_list.append(student)

print("\n----- Final Results -----")
for s in students_list:
    print("Name:", s['name'], " Roll:", s['roll'], " Total:", s['total'], "Per:", s['percentage'], " Grade:", s['grade'])

highest = students_list[0]
for s in students_list:
    if s['percentage'] > highest['percentage']:
        highest = s

print("\nTopper is", highest['name'], "with", highest['percentage'], "%")


file = open("results.txt", "w")

for s in students_list:
    file.write(s['name'])
    file.write(", ")
    file.write(s['roll'])
    file.write(", Total=")
    file.write(str(s['total']))
    file.write(", Per=")
    file.write(str(s['percentage']))
    file.write(", Grade=")
    file.write(s['grade'])
    file.write("\n")

file.write("\nTopper: ")
file.write(highest['name'])
file.write("\n")
file.close()

print("File results.txt saved successfully!")
```

```python
# Q2 Sentence Analysis

sentence = input("Enter any sentence: ")
words = sentence.split()

total_words = len(words)
total_chars = len(sentence)

longest = words[0]
shortest = words[0]

for w in words:
    if len(w) > len(longest):
        longest = w
    if len(w) < len(shortest):
        shortest = w

vowels = 0
for ch in sentence:
    if ch in "aeiouAEIOU":
        vowels = vowels + 1

unique_words = set(words)
alpha_order = sorted(words)

print("\nTotal words:", total_words)
print("Total chars:", total_chars)
print("Longest word:", longest)
print("Shortest word:", shortest)
print("Vowels:", vowels)
print("Unique words:", unique_words)
print("Alphabetical:", alpha_order)

f = open("word_analysis.txt", "w")
f.write("Sentence: " + sentence + "\n")
f.write("Total words: " + str(total_words) + "\n")
f.write("Total chars: " + str(total_chars) + "\n")
f.write("Longest: " + longest + "\n")
f.write("Shortest: " + shortest + "\n")
f.write("Vowels: " + str(vowels) + "\n")
f.write("Unique: " + str(unique_words) + "\n")
f.write("Sorted: " + str(alpha_order) + "\n")
f.close()
```

```python
# Q3 Shopping Program

def calculate_final_bill(products):
    total = 0
    for p in products:
        total = total + p['total_price']

    discount = 0
    if total > 10000:
        discount = total * 0.10

    final = total - discount
    return total, discount, final

products = []

for i in range(5):
    print("\nProduct", i+1)
    name = input("Name: ")
    price = float(input("Price: "))
    qty = int(input("Quantity: "))
    total_price = price * qty

    d = {"name": name, "price": price, "quantity": qty, "total_price": total_price}
    products.append(d)

total_bill, discount, final_bill = calculate_final_bill(products)

expensive = products[0]
for p in products:
    if p['price'] > expensive['price']:
        expensive = p

print("\n--- BILL ---")
for p in products:
    print(p['name'], "=", p['price'], "x", p['quantity'], "=", p['total_price'])

print("Total Bill:", total_bill)
print("Discount:", discount)
print("Final Bill:", final_bill)
print("Most Expensive:", expensive['name'])

file = open("bill.txt", "w")
file.write("--- Shopping Bill ---\n")
for p in products:
    file.write(p['name'] + " - " + str(p['total_price']) + "\n")
file.write("Total: " + str(total_bill) + "\n")
file.write("Discount: " + str(discount) + "\n")
file.write("Final: " + str(final_bill) + "\n")
file.close()
```

```python
# Q4 Calculator

def add(a,b):
    return a+b

def sub(a,b):
    return a-b

def multiply(a,b):
    return a*b

def divide(a,b):
    if b == 0:
        return "Cannot divide by zero"
    return a/b

def modulus(a,b):
    return a%b

while True:
    n1 = float(input("\nEnter first number: "))
    n2 = float(input("Enter second number: "))

    print("Operations: + - * / %")
    op = input("Enter operation: ")

    if op == '+':
        print("Result:", add(n1,n2))
    elif op == '-':
        print("Result:", sub(n1,n2))
    elif op == '*':
        print("Result:", multiply(n1,n2))
    elif op == '/':
        print("Result:", divide(n1,n2))
    elif op == '%':
        print("Result:", modulus(n1,n2))
    else:
        print("Wrong operation!")

    again = input("Do you want to calculate again? (yes/no): ")
    if again!= "yes":
        print("Bye Bye Calculator closed")
        break
```

```python
# Q5 Restaurant Program

menu = {
    1: {"name": "Burger", "price": 500},
    2: {"name": "Pizza", "price": 800},
    3: {"name": "Fries", "price": 250},
    4: {"name": "Drink", "price": 150}
}

print("--- MENU ---")
print("1. Burger Rs.500")
print("2. Pizza Rs.800")
print("3. Fries Rs.250")
print("4. Drink Rs.150")

orders = []
total_bill = 0

while True:
    choice = int(input("\nEnter item number (0 to stop): "))
    if choice == 0:
        break

    if choice in menu:
        qty = int(input("How many " + menu[choice]['name'] + "? "))
        price = menu[choice]['price'] * qty
        total_bill = total_bill + price
        orders.append(menu[choice]['name'] + " x " + str(qty) + " = Rs." + str(price))
        print("Added!")
    else:
        print("Invalid item!")

print("\nYour Order:")
for o in orders:
    print(o)

print("Total Bill: Rs.", total_bill)

f = open("orders.txt", "w")
f.write("Restaurant Orders\n")
for o in orders:
    f.write(o + "\n")
f.write("Final Bill: " + str(total_bill))
f.close()
```
